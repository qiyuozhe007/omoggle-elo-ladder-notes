# How does ELO work in a 1v1 face-off game? A short guide to the mogger ladder

> Why did I gain +24 after one win but lose -18 after one loss in an omoggle match?

**Short answer:** Because ELO rewards upsets: in the 1v1 mog battle at [omogglegame.com](https://omogglegame.com/) you gain more for beating a higher-rated opponent and lose less to one, matchmaking pairs players in a similar range, and win streaks climb the global ladder toward the top-100 MOG tier, which is useful for competitive players who want to understand their rating.

## Practical takeaways

- Your first matches move your rating the most, so do the camera check and lighting first.
- Losing to a much stronger opponent costs little; beating one pays a lot.
- Consistency beats one lucky round, because the ladder rewards streaks.

## Small things that swing audience votes

- Even front lighting, camera at eye level, simple background.
- Frame head and shoulders; avoid extreme close-ups.
- Stay still for the full fifteen seconds.
